# Introdution

Until now, we've looked at the fundamentals of preprocessing and the use of the bag-of-words approach to model topics in LDA (Latent Dirichlet Allocation).

Now, let's expand on some concepts and revisit, for example, the use of the bag-of-words approach to compare documents. Imagine that we have two documents, and we want to determine whether they discuss the same topic, based on the assumption that if both documents share the same terms, they likely cover the same subject.

With this in mind, we will use the concept of document vectors in the bag-of-words model, which can be explained as representing each document as a vector of word counts or frequencies within a fixed vocabulary. In this vector representation, each dimension corresponds to a specific word in the vocabulary, and the value in each dimension represents how often that word appears in the document.



# Vectors

By transforming documents into these numerical vectors, we can easily compare them using similarity metrics, such as cosine similarity. If two document vectors have high similarity, this suggests that the documents share a significant amount of overlapping content, indicating a related subject. The bag-of-words approach thus provides a straightforward way to quantify the content of documents for comparisons and further analysis in natural language processing tasks.

To compare documents using the Bag of Words approach with CountVectorizer from scikit-learn, we can follow these steps:

1. Transform each document into a word count vector using CountVectorizer.
2. Calculate the similarity between the vectors. One of the most common methods is cosine similarity.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Exemplo de documentos
doc1 = "Eu amo aprender sobre processamento de linguagem natural"
doc2 = "A pipoca é deliciosa"

# Lista de documentos
documents = [doc1, doc2]

# Inicializa o CountVectorizer
vectorizer = CountVectorizer()

# Converte os documentos para a representação bag-of-words
X = vectorizer.fit_transform(documents)

# Calcula a similaridade do cosseno entre os vetores dos documentos
cosine_sim = cosine_similarity(X[0], X[1])

# Exibe a similaridade entre doc1 e doc2
print("Similaridade entre doc1 e doc2:", cosine_sim[0][0])
```

Here is an example with three documents: two on the same topic (natural language processing) and a third on a different subject (weather). Using the above code calculate the cosine similarity.

```python
# Exemplo de documentos
doc1 = "O processamento de linguagem natural é fascinante e cheio de possibilidades."
doc2 = "A área de processamento de linguagem natural envolve a análise de textos e fala."
doc3 = "Hoje o clima está ensolarado com algumas nuvens."

from nltk.corpus import stopwords

# Lista de documentos
# documents = [doc1.lower(), doc2.lower(), doc3.lower()]
documents = [doc1, doc2, doc3]

# Inicializa o CountVectorizer
vectorizer = CountVectorizer(lowercase=True,
                             token_pattern=r"\b\w+\b")
                             # stop_words=stopwords.words("portuguese"))

# Converte os documentos para a representação bag-of-words
X = vectorizer.fit_transform(documents)

# print vocabulary
print("Vocabulary:", vectorizer.get_feature_names_out())
print(len(vectorizer.get_feature_names_out()))

# Calcula a similaridade do cosseno entre os documentos
cosine_sim = cosine_similarity(X)
print(cosine_sim)

# Exibe a similaridade entre os documentos
print("Similaridade entre doc1 e doc2 (mesmo tema):", cosine_sim[0][1])
print("Similaridade entre doc1 e doc3 (temas diferentes):", cosine_sim[0][2])
print("Similaridade entre doc2 e doc3 (temas diferentes):", cosine_sim[1][2])
```

In what range does cosine similarity work?

```python
#The cosine similarity range is from -1 to 1. However, in the context of bag-of-words models, cosine similarity values typically range
#from 0 to 1, since word count vectors are generally non-negative. A value of 1 indicates that the vectors are identical in terms of direction,
#meaning they share the same distribution of terms. A cosine similarity of 0 implies no similarity, suggesting that the vectors are orthogonal,
#or that they have no terms in common. Negative values, which represent opposite directions, are rare in bag-of-words applications,
#as the vectors do not usually contain negative components.
```

In the example above, we used cosine similarity as a metric, but there are other alternatives that can be employed to compare vectors generated by the bag-of-words model. Some of these include Euclidean distance, which measures the straight-line distance between two points in a multi-dimensional space, and Manhattan distance, which calculates the sum of the absolute differences across each dimension. Additionally, the Jaccard similarity coefficient can be used to measure the overlap between two sets, making it particularly useful when focusing on the presence or absence of words rather than their frequency. Each metric has its own advantages depending on the context and desired outcome of the comparison, as some are more sensitive to vector magnitude, while others emphasize the direction or relative frequency of terms.

# Week 6 — Sentiment Analysis and Text Categorization

In this lab, we classify app reviews as **negative (0)** or **positive (1)** and compare two approaches:

1. **VADER**, a well-known English sentiment lexicon with rules for cues such as negation, emphasis, and punctuation. It does not learn from this dataset.
2. **Multinomial Naive Bayes**, which learns associations between words and labels from annotated training reviews.

We load the dataset once and create one shared train/test split. Both approaches will be evaluated on the **same test reviews and labels**. VADER may leave some reviews neutral, so we will also report its coverage.


## Shared dataset and train/test split

The CSV contains the review text and its human-assigned polarity. We keep the same rows for both approaches, remove rows missing either field once, and split the data once. We preserve the original review text so VADER can use punctuation and capitalization cues. `CountVectorizer` will apply its own lowercasing when it builds Naive Bayes features.

Run this cell once. The variables `reviews`, `X_train`, `X_test`, `y_train`, and `y_test` are shared below by the lexicon and Naive Bayes sections.


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Load the CSV exactly once and prepare the shared set of labeled reviews.
data = pd.read_csv("google_play_store_apps_reviews_training.csv")
reviews = data[["review", "polarity"]].dropna().copy()
reviews["review"] = reviews["review"].astype(str).str.strip()

# Create one common held-out test set for both approaches.
X_train, X_test, y_train, y_test = train_test_split(
    reviews["review"],
    reviews["polarity"],
    test_size=0.25,
    random_state=42,
    stratify=reviews["polarity"],
)

print("Shared labeled reviews:", len(reviews))
print("Training reviews:", len(X_train))
print("Shared test reviews:", len(X_test))
print("Labels (0=negative, 1=positive):")
print(reviews["polarity"].value_counts().sort_index())
reviews.head()
```

## 1 — Lexicon-based sentiment with VADER

VADER (Valence Aware Dictionary and sEntiment Reasoner) is a well-known English lexicon and rule-based sentiment method. Unlike a plain word-score sum, VADER also adjusts scores for common cues such as negation, intensifiers, capitalization, and punctuation. These rules help with some context, but they do not provide a full understanding of meaning or sarcasm.

NLTK includes `SentimentIntensityAnalyzer`. Its `polarity_scores()` method returns `neg`, `neu`, `pos`, and `compound`. The normalized `compound` score ranges from −1 (most negative) to +1 (most positive). For this binary dataset, we will use the commonly documented thresholds: compound ≥ 0.05 means positive, compound ≤ −0.05 means negative, and values between these thresholds remain neutral/unclassified. See the [NLTK VADER documentation](https://www.nltk.org/api/nltk.sentiment.vader.html) and [VADER project documentation](https://github.com/cjhutto/vaderSentiment#about-the-scoring).

If NLTK is not installed in your environment, install it before running the next cell with `%pip install nltk`.


```python
import nltk
nltk.download("vader_lexicon")

from nltk.sentiment.vader import SentimentIntensityAnalyzer

vader = SentimentIntensityAnalyzer()
print("VADER lexicon entries:", len(vader.lexicon))
```

### Inspect some VADER lexicon entries

The lexicon maps individual terms to valence scores. Positive values indicate positive valence and negative values indicate negative valence. We inspect a few familiar entries before applying the analyzer to full reviews.


```python
sample_terms = ["great", "good", "excellent", "love", "bad", "terrible", "slow", "hate"]
vader_lexicon_examples = pd.DataFrame([
    {"term": term, "valence": vader.lexicon[term]}
    for term in sample_terms
    if term in vader.lexicon
])
vader_lexicon_examples
```

## Calculating the Compound Score in VADER

VADER (Valence Aware Dictionary and sEntiment Reasoner) calculates an aggregated sentiment score called the **Compound Score**. This score represents the overall emotional intensity and polarity of a text, ranging from **-1 (extremely negative) to +1 (extremely positive)**.

The calculation takes place in two main steps:

1. **Valence aggregation:** Sum of the emotional valence scores of the words, after applying contextual adjustments such as negation, intensifiers, capitalization, and punctuation.
2. **Normalization:** Transformation of the aggregated valence into a score bounded between -1 and +1.

### 1. Normalization formula

VADER uses the following formula:

$$
compound = \frac{S}{\sqrt{S^2 + 15}}
$$

Where:

- $S$: Sum of the adjusted valence scores of the words in the text.
- $15$: Normalization constant, empirically selected by the authors of VADER.
- $\sqrt{S^2 + 15}$: Denominator that scales the result to a range between -1 and +1.

The constant 15 controls how quickly the result approaches the extremes of the scale. It does not represent the number of words or the document length.

### 2. Calculation example

Consider the following sentence:

"The movie is good and fantastic."

For simplicity, let us consider only the words with emotional content:

| Word | Valence |
|---|---:|
| good | +1.9 |
| fantastic | +3.0 |

**Step 1 — Valence aggregation**

$$
S = 1.9 + 3.0 = 4.9
$$

**Step 2 — Applying the formula**

Replacing $S$ with 4.9:

$$
compound = \frac{4.9}{\sqrt{4.9^2 + 15}}
$$

Calculating the square:

$$
compound = \frac{4.9}{\sqrt{24.01 + 15}}
$$

Adding the values:

$$
compound = \frac{4.9}{\sqrt{24.01 + 15}}
$$

Calculating the square root:

$$
compound = \frac{4.9}{\sqrt{39.01}}
$$

Approximating the square root:

$$
compound = \frac{4.9}{6.245}
$$

Finally, dividing:

$$
compound \approx 0.785
$$


### Apply VADER to the shared test reviews

We score `X_test`, the same held-out reviews that Naive Bayes will later classify. Reviews inside the neutral band receive no binary prediction. We therefore report both **coverage** (the share assigned positive or negative) and **accuracy among covered reviews**.


```python
X_test
```

```python

# Apply VADER sentiment analysis to exactly the same X_test observations used
# in the Naive Bayes evaluation, ensuring a fair comparison between models.
vader_components = X_test.apply(vader.polarity_scores).apply(pd.Series)
# vader_components

# Combine the original reviews, their true sentiment labels (y_test),
# and the four VADER sentiment scores (neg, neu, pos, compound) into one DataFrame.
vader_results = pd.concat(
    [X_test.rename("review"), y_test.rename("polarity"), vader_components], axis=1
)

# Create a new column to store VADER's binary sentiment predictions.
# Initialize all predictions as missing (pd.NA), using the nullable integer type.
vader_results["vader_prediction"] = pd.Series(pd.NA, index=X_test.index, dtype="Int64")

# Classify reviews with a compound score >= 0.05 as positive (label 1).
vader_results.loc[vader_results["compound"] >= 0.05, "vader_prediction"] = 1

# Classify reviews with a compound score <= -0.05 as negative (label 0).
vader_results.loc[vader_results["compound"] <= -0.05, "vader_prediction"] = 0

# Identify reviews for which VADER produced a binary prediction.
# Reviews with compound scores between -0.05 and +0.05 are considered neutral
# and are excluded from the binary classification evaluation.
vader_covered = vader_results["vader_prediction"].notna()

# Calculate VADER coverage: the proportion of test reviews assigned
# a positive or negative prediction (excluding neutral reviews).
vader_coverage = vader_covered.mean()

# Display the percentage of test reviews covered by VADER.
print(f"VADER coverage on shared test set: {vader_coverage:.1%}")

# Check whether VADER produced at least one binary prediction.
if vader_covered.any():

    # Calculate classification accuracy using only the reviews
    # for which VADER generated a positive or negative prediction.
    # Compare the true labels (y_test) against VADER's predictions.
    vader_accuracy = accuracy_score(
        y_test.loc[vader_covered],
        vader_results.loc[vader_covered, "vader_prediction"].astype(int),
    )

    # Display VADER's accuracy on the covered reviews.
    print(f"VADER accuracy on covered reviews: {vader_accuracy:.3f}")

    # Display the confusion matrix.
    # Rows represent the true labels and columns represent predicted labels.
    # The order [0, 1] corresponds to negative and positive sentiment, respectively.
    print("Confusion matrix (rows=true, columns=predicted; labels 0, 1):")

    print(confusion_matrix(
        y_test.loc[vader_covered],
        vader_results.loc[vader_covered, "vader_prediction"].astype(int),
        labels=[0, 1],
    ))

# Display the first five rows of the resulting DataFrame,
# including the original reviews, true labels, VADER scores, and predictions.
vader_results.head()
```

**Interpretation.** VADER assigns no positive/negative label to reviews whose compound score falls in the neutral band. Its accuracy therefore describes only the covered part of the shared test set. Always read accuracy together with coverage.

```python
#classification report
if vader_covered.any():
    print("\nClassification report:")
    print(classification_report(
        y_test.loc[vader_covered],
        vader_results.loc[vader_covered, "vader_prediction"].astype(int),
        target_names=["negative (0)", "positive (1)"],
        zero_division=0
    ))
```

## 2 — Text categorization with Multinomial Naive Bayes

Text categorization assigns one of a set of labels to a document. Here, each review is a document and its polarity is the label. `CountVectorizer` converts the training reviews into a bag-of-words matrix: each column represents a vocabulary term and each cell records its count in a review.

Multinomial Naive Bayes estimates how likely each word is under each class, then combines these likelihoods with the class prior. Its simplifying assumption is that terms contribute independently given the class. The model learns from the training reviews and predicts labels for the same held-out `X_test` used by VADER.

https://www.youtube.com/watch?v=O2L2Uv9pdDA&t=639s

### Inspect the shared split

The shared setup cell already created the train/test split with stratification. We do not split or reload the data again. `X_train` and `y_train` are used to fit Naive Bayes; `X_test` and `y_test` are reserved for evaluation.


```python
print("Train labels:")
print(y_train.value_counts().sort_index())
print("\nTest labels:")
print(y_test.value_counts().sort_index())
print("\nThe lexicon and Naive Bayes test sets have the same indices:", X_test.index.equals(y_test.index))
```

### Convert the shared reviews into count vectors

Fit the vectorizer only on `X_train`. This defines a vocabulary without looking at the test set. Then use `transform()` on `X_test` to apply that same vocabulary to the test reviews.


```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

vectorizer = CountVectorizer(stop_words="english", min_df=2)
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

print("Training matrix (reviews, terms):", X_train_vec.shape)
print("Test matrix (reviews, terms):", X_test_vec.shape)
print("First 20 vocabulary terms:", vectorizer.get_feature_names_out()[:20])
```

### Train Naive Bayes and evaluate on the shared test set

Fit the classifier with the training vectors and labels. Then predict `X_test`, which is identical to the test split scored by VADER. We report Naive Bayes accuracy on the whole test set and also on VADER's covered subset, making a direct comparison possible for the same reviews.


```python

# Initialize the Multinomial Naive Bayes classifier.
# This algorithm is commonly used for text classification because it works
# well with discrete features such as word frequencies or TF-IDF values.
model = MultinomialNB()

# Train the Naive Bayes model using the vectorized training documents (X_train_vec)
# and their corresponding sentiment labels (y_train).
model.fit(X_train_vec, y_train)

# Use the trained model to predict the sentiment of the test documents.
# Convert the predictions into a Pandas Series, preserving the original
# X_test index to ensure correct alignment with the other results.
# Assign the column name "nb_prediction" to identify the predictions.
y_pred = pd.Series(
    model.predict(X_test_vec),
    index=X_test.index,
    name="nb_prediction"
)

# Calculate the classification accuracy across all test reviews.
# Accuracy represents the proportion of correctly predicted sentiment labels.
nb_accuracy = accuracy_score(y_test, y_pred)

# Display the Naive Bayes accuracy on the complete shared test set.
# The :.3f format displays the result with three decimal places.
print(f"Naive Bayes accuracy on all shared test reviews: {nb_accuracy:.3f}")

# Display a description of the confusion matrix.
# Rows represent the actual sentiment labels and columns represent predictions.
print("Confusion matrix (rows=true, columns=predicted; labels 0, 1):")

# Generate the confusion matrix using the true labels (y_test)
# and the predicted labels (y_pred).
# The labels=[0, 1] argument ensures that:
# 0 = negative sentiment
# 1 = positive sentiment
print(confusion_matrix(y_test, y_pred, labels=[0, 1]))

# Print a heading before displaying the classification report.
print("\nClassification report:")

# Generate a detailed classification report containing:
# - Precision: proportion of positive predictions that are correct.
# - Recall: proportion of actual class instances correctly identified.
# - F1-score: harmonic mean of precision and recall.
# - Support: number of actual observations belonging to each class.
#
# target_names assigns readable names to the numerical labels:
# 0 = negative sentiment
# 1 = positive sentiment
#
# zero_division=0 avoids division-by-zero warnings and assigns a score of 0
# when a metric cannot be calculated because there are no relevant predictions.
print(classification_report(
    y_test,
    y_pred,
    target_names=["negative (0)", "positive (1)"],
    zero_division=0
))

# Check whether VADER generated at least one binary prediction.
# This condition ensures that the following comparison is only performed
# when there are reviews covered by VADER.
if vader_covered.any():

    # Calculate Naive Bayes accuracy using only the reviews for which
    # VADER also produced a positive or negative prediction.
    #
    # This allows a fair comparison between the two approaches
    # by evaluating both models on exactly the same subset of reviews.
    nb_accuracy_on_vader_coverage = accuracy_score(
        y_test.loc[vader_covered],
        y_pred.loc[vader_covered]
    )

    # Display Naive Bayes accuracy on the subset of reviews covered by VADER.
    # This result can be directly compared with VADER's accuracy
    # calculated on the same subset of test reviews.
    print(f"Naive Bayes accuracy on VADER-covered reviews: {nb_accuracy_on_vader_coverage:.3f}")
```

**Compare the methods.** Both methods now use the same test reviews and human labels. Compare their accuracy on the subset VADER covers. Then consider the coverage difference: Naive Bayes always predicts one of the two classes, while VADER can leave neutral-band cases unclassified. A lexicon is fixed and inspectable; Naive Bayes learns word/class associations from the labeled training data.

### Predict new reviews with Naive Bayes

For new text, use the existing vectorizer and model. Do not fit a new vocabulary. The probabilities are model estimates, not guaranteed calibrated confidence values.


```python

# Define new reviews to be classified by the trained Naive Bayes model.
new_reviews = [
    "Application is perfect with all the functions you need. I will always recommend it.",
    "I bought it to use as a digital assistant. It doesn't work. And there's no information about it.",
]

# Transform the new reviews using the original fitted vectorizer.
# This ensures that the new documents have exactly the same features
# (vocabulary and feature dimensions) used during model training.
new_vectors = vectorizer.transform(new_reviews)

# Predict the sentiment class for each new review.
# 0 = negative sentiment
# 1 = positive sentiment
new_predictions = model.predict(new_vectors)

# Calculate the estimated probability of each sentiment class.
# The output contains one probability for each class:
# [P(negative), P(positive)].
new_probabilities = model.predict_proba(new_vectors)

# Iterate simultaneously over the original reviews, predicted labels,
# and estimated class probabilities.
for review, prediction, probabilities in zip(
    new_reviews, new_predictions, new_probabilities
):
    # Convert the numerical prediction into a readable sentiment label.
    label = "positive" if prediction == 1 else "negative"

    # Display the original review and its predicted sentiment.
    print(f"Review: {review}\nPrediction: {label} ({prediction})")

    # Display the estimated probabilities for negative and positive sentiment.
    print(f"Estimated probabilities [negative, positive]: {probabilities}\n")
```
